```python
DATA_PATH = "../data/raw/Fraud.csv"

df = pd.read_csv(DATA_PATH)

print(f"Dataset loaded successfully.")
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

```python
schema = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "non_null_count": df.notna().sum().values,
    "missing_count": df.isna().sum().values,
    "unique_values": df.nunique().values
})

schema
```

```python
missing_values = (
    df.isna()
      .sum()
      .rename("missing_count")
      .to_frame()
)

missing_values["missing_percentage"] = (
    missing_values["missing_count"] / len(df) * 100
)

missing_values
```

```python
total_missing = df.isna().sum().sum()

print(f"Total missing values: {total_missing:,}")
```

```python
duplicate_count = df.duplicated().sum()
duplicate_percentage = duplicate_count / len(df) * 100

print(f"Duplicate rows: {duplicate_count:,}")
print(f"Duplicate percentage: {duplicate_percentage:.6f}%")
```

```python
numerical_summary = df.describe().T

numerical_summary
```

```python
total_transactions = len(df)
fraud_transactions = int(df["isFraud"].sum())
legitimate_transactions = total_transactions - fraud_transactions

fraud_rate = fraud_transactions / total_transactions
legitimate_rate = legitimate_transactions / total_transactions

print(f"Total transactions:      {total_transactions:,}")
print(f"Fraudulent transactions: {fraud_transactions:,}")
print(f"Legitimate transactions: {legitimate_transactions:,}")
print(f"Fraud rate:              {fraud_rate:.6%}")
print(f"Legitimate rate:         {legitimate_rate:.6%}")
```

```python
total_transaction_value = df["amount"].sum()

fraudulent_transaction_value = df.loc[
    df["isFraud"] == 1,
    "amount"
].sum()

legitimate_transaction_value = df.loc[
    df["isFraud"] == 0,
    "amount"
].sum()

average_fraud_amount = df.loc[
    df["isFraud"] == 1,
    "amount"
].mean()

average_legitimate_amount = df.loc[
    df["isFraud"] == 0,
    "amount"
].mean()

fraud_value_percentage = (
    fraudulent_transaction_value / total_transaction_value
)

print(f"Total transaction value:       {total_transaction_value:,.2f}")
print(f"Fraudulent transaction value:  {fraudulent_transaction_value:,.2f}")
print(f"Legitimate transaction value:  {legitimate_transaction_value:,.2f}")
print(f"Average fraudulent amount:     {average_fraud_amount:,.2f}")
print(f"Average legitimate amount:     {average_legitimate_amount:,.2f}")
print(f"Fraudulent share of value:     {fraud_value_percentage:.6%}")
```

```python
median_fraud_amount = df.loc[
    df["isFraud"] == 1,
    "amount"
].median()

median_legitimate_amount = df.loc[
    df["isFraud"] == 0,
    "amount"
].median()

print(f"Median fraudulent amount: {median_fraud_amount:,.2f}")
print(f"Median legitimate amount: {median_legitimate_amount:,.2f}")
```

```python
minimum_step = df["step"].min()
maximum_step = df["step"].max()
number_of_steps = df["step"].nunique()

print(f"Minimum step:    {minimum_step}")
print(f"Maximum step:    {maximum_step}")
print(f"Unique steps:    {number_of_steps}")
print(f"Approximate days represented: {number_of_steps / 24:.2f}")
```

```python
transactions_by_type = (
    df.groupby("type", observed=True)
      .agg(
          transaction_count=("type", "size"),
          total_value=("amount", "sum"),
          average_amount=("amount", "mean"),
          median_amount=("amount", "median")
      )
      .sort_values("transaction_count", ascending=False)
)

transactions_by_type["transaction_percentage"] = (
    transactions_by_type["transaction_count"]
    / total_transactions
    * 100
)

transactions_by_type
```

```python
df_analysis = df.assign(
    fraudulent_amount=df["amount"] * df["isFraud"]
)

fraud_summary_by_type = (
    df_analysis.groupby("type", observed=True)
      .agg(
          transaction_count=("type", "size"),
          fraud_count=("isFraud", "sum"),
          total_value=("amount", "sum"),
          fraudulent_value=("fraudulent_amount", "sum")
      )
)

fraud_summary_by_type["fraud_rate"] = (
    fraud_summary_by_type["fraud_count"]
    / fraud_summary_by_type["transaction_count"]
)

fraud_summary_by_type["fraud_share_of_value"] = (
    fraud_summary_by_type["fraudulent_value"]
    / fraud_summary_by_type["total_value"]
)

fraud_summary_by_type = fraud_summary_by_type.sort_values(
    "fraud_count",
    ascending=False
)

fraud_summary_by_type
```

```python
fraud_summary_display = fraud_summary_by_type.copy()

fraud_summary_display["fraud_rate"] = (
    fraud_summary_display["fraud_rate"] * 100
)

fraud_summary_display["fraud_share_of_value"] = (
    fraud_summary_display["fraud_share_of_value"] * 100
)

fraud_summary_display = fraud_summary_display.rename(columns={
    "transaction_count": "Transactions",
    "fraud_count": "Frauds",
    "total_value": "Total Value",
    "fraudulent_value": "Fraudulent Value",
    "fraud_rate": "Fraud Rate (%)",
    "fraud_share_of_value": "Fraud Share of Value (%)"
})

fraud_summary_display
```

## Initial Findings

1. Fraud represents only 0.1291% of transactions but approximately
   1.05% of transaction value, demonstrating that fraud is rare but
   financially disproportionate.

2. The average fraudulent transaction is approximately 8.2 times
   larger than the average legitimate transaction. However, amount
   alone cannot establish whether a transaction is fraudulent.

3. Fraud occurs only in TRANSFER and CASH_OUT transactions within
   PaySim. TRANSFER has approximately 4.2 times the fraud rate of
   CASH_OUT, while the two types contribute nearly equal fraudulent value.

4. The existing rule achieves 100% precision but only approximately
   0.19% recall, indicating that it produces highly reliable alerts but
   misses almost all fraudulent transactions.

5. The dataset contains no missing values or exact duplicate rows.
   This establishes structural completeness but does not guarantee
   transactional validity or real-world representativeness.
